```python
'''~
이 코드는 평균이 1이고 분산이 동일한 두 확률 변수 X, Y의 2차원 결합 확률 분포를 생성하고,
이를 Z = X + Y 라는 새로운 선형 축으로 정사영(Projection)했을 때의 기하학적 중심 변환을 시각화합니다.
[이전 대화 대비 달라진 점]
- 추상적인 3차원 투영 대신, 직관적인 2차원 평면(등고선)과 1차원 히스토그램을 대비시켜 기댓값의 선형성(1+1=2)을 증명하도록 설계했습니다.
- 요청하신 대로 최상단에 코드의 목적과 변경점을 명시하였으며, 내부 주석은 간결한 한국어로 작성했습니다.
~'''

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정 (환경에 따라 수정 필요)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 1. 데이터 생성 (평균 1, 표준편차 1)
np.random.seed(42)
n_samples = 30000
X = np.random.normal(1, 1, n_samples)
Y = np.random.normal(1, 1, n_samples)
Z = X + Y

# 2. 그래프 시각화 설정
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 첫 번째 그래프: 2차원 결합 확률 분포 (X, Y)
ax1 = axes[0]
sns.kdeplot(x=X, y=Y, ax=ax1, cmap="Blues", fill=True, alpha=0.5)
ax1.scatter(X, Y, s=2, color='blue', alpha=0.1)

# 무게중심 (1, 1) 표시
ax1.plot(1, 1, 'ro', markersize=10, label='무게중심 (1, 1)')

# Z = X + Y 투영선 (x + y = z -> y = -x + z)
x_vals = np.linspace(-2, 4, 100)
ax1.plot(x_vals, -x_vals + 2, 'r--', linewidth=2, label='Z=2 투영선 (x+y=2)')
ax1.plot(x_vals, -x_vals + 0, 'g:', linewidth=1.5, label='Z=0 투영선 (x+y=0)')
ax1.plot(x_vals, -x_vals + 4, 'y:', linewidth=1.5, label='Z=4 투영선 (x+y=4)')

ax1.set_title('2차원 평면에서의 결합 분포와 정사영 축')
ax1.set_xlabel('X 분포')
ax1.set_ylabel('Y 분포')
ax1.set_xlim(-2, 4)
ax1.set_ylim(-2, 4)
ax1.axhline(0, color='black', linewidth=0.5)
ax1.axvline(0, color='black', linewidth=0.5)
ax1.legend()

# 두 번째 그래프: 1차원 정사영 분포 (Z = X + Y)
ax2 = axes[1]
sns.histplot(Z, kde=True, ax=ax2, color='purple', bins=40)

# 새로운 기댓값 Z=2 표시
ax2.axvline(2, color='red', linestyle='--', linewidth=3, label='새로운 기댓값 E[Z] = 2')

ax2.set_title('Z = X + Y 로 투영된 1차원 분포')
ax2.set_xlabel('Z 분포 (X + Y)')
ax2.set_ylabel('빈도 (확률 밀도)')
ax2.legend()

plt.tight_layout()
plt.show()
```
