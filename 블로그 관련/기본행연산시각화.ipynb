```python
'''
기본 행 연산에 따른 선형계의 기하학적 변환 과정을 시각화하는 파이썬 코드입니다.
이전 대화에서 분석한 두 직선 x + y = 1 및 -2x + y = 1의 교점(0, 1)이 유지된 상태로
좌표축에 직교하도록 회전하는 대수적·기하학적 단계를 3개의 서브플롯으로 구성했습니다.
'''

import numpy as np
import matplotlib.pyplot as plt

# x 좌표 공간 정의
x = np.linspace(-3, 3, 400)

# 시각화 환경 설정
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
intersection = (0, 1)

# 단계 1: 초기 선형계
# L1: x + y = 1 -> y = -x + 1
# L2: -2x + y = 1 -> y = 2x + 1
y1_init = -x + 1
y2_init = 2*x + 1

axs[0].plot(x, y1_init, label='L1: x + y = 1', color='blue')
axs[0].plot(x, y2_init, label='L2: -2x + y = 1', color='red')
axs[0].set_title('Initial State')

# 단계 2: R2 <- R2 + 2R1 및 R2 <- (1/3)R2 연산 후
# L1: x + y = 1 -> y = -x + 1
# L2: y = 1 (기하학적 위치 동일)
y1_mid = -x + 1
y2_mid = np.ones_like(x)

axs[1].plot(x, y1_mid, label='L1: x + y = 1', color='blue')
axs[1].plot(x, y2_mid, label='L2: y = 1', color='red')
axs[1].set_title('After R2 Operations')

# 단계 3: R1 <- R1 - R2 연산 후 (기약 행 사다리꼴)
# L1: x = 0
# L2: y = 1
y2_final = np.ones_like(x)

axs[2].axvline(x=0, label='L1: x = 0', color='blue')
axs[2].plot(x, y2_final, label='L2: y = 1', color='red')
axs[2].set_title('Final State (RREF)')

# 공통 설정 적용
for ax in axs:
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_xlim(-3, 3)
    ax.set_ylim(-2, 4)
    ax.grid(True, linestyle='--', alpha=0.6)
    # 교점 표시
    ax.plot(intersection[0], intersection[1], 'ko', markersize=6, label='Intersection (0,1)')
    ax.legend(loc='upper right')

plt.tight_layout()
plt.show()
```
